````python
# AI attempt


from datetime import datetime
from dataclasses import dataclass, field
from typing import Optional
import uuid


@dataclass
class Transaction:
    """
    A reusable blueprint for a financial transaction.
    """

    amount: float
    category: str
    description: str
    date: datetime = field(default_factory=datetime.now)

    account: Optional[str] = None
    merchant: Optional[str] = None

    # auto-generated unique ID
    transaction_id: str = field(default_factory=lambda: str(uuid.uuid4()))

    # -------------------------
    # Basic helper methods
    # -------------------------

    def is_expense(self) -> bool:
        return self.amount < 0

    def is_income(self) -> bool:
        return self.amount > 0

    def update_amount(self, new_amount: float):
        self.amount = new_amount

    def update_category(self, new_category: str):
        self.category = new_category

    def update_description(self, new_description: str):
        self.description = new_description

    def summary(self) -> str:
        sign = "-" if self.is_expense() else "+"
        return (
            f"[{self.transaction_id}] {self.date.strftime('%Y-%m-%d')} | "
            f"{self.description} | {sign}${abs(self.amount):.2f} | "
            f"{self.category}"
        )
````

````python
from dataclasses import dataclass, field
from datetime import datetime
import uuid


# -------------------------
# 1. Transaction Template
# -------------------------
@dataclass
class Transaction:
    date: datetime
    description: str
    amount: float
    category: str

    transaction_id: str = field(default_factory=lambda: str(uuid.uuid4()))

    # -------------------------
    # Business rules
    # -------------------------
    def is_expense(self):
        return self.amount < 0

    def is_income(self):
        return self.amount > 0

    def is_large_expense(self):
        return self.is_expense() and abs(self.amount) > 50

    def is_this_month(self):
        now = datetime.now()
        return self.date.year == now.year and self.date.month == now.month


# -------------------------
# 2. Sample data (spreadsheet-style)
# -------------------------
transaction_data = [
    {"date": "2025-09-20", "description": "Starbucks Coffee", "amount": -4.50, "category": "Food"},
    {"date": "2025-09-21", "description": "Salary Deposit", "amount": 3000.00, "category": "Income"},
    {"date": "2025-09-22", "description": "Grocery Shopping", "amount": -67.85, "category": "Food"},
    {"date": "2025-09-23", "description": "Gas Station", "amount": -45.00, "category": "Transport"}
]


# -------------------------
# 3. Convert to objects
# -------------------------
transactions = [
    Transaction(
        date=datetime.strptime(row["date"], "%Y-%m-%d"),
        description=row["description"],
        amount=row["amount"],
        category=row["category"]
    )
    for row in transaction_data
]


# -------------------------
# 4. TABLE OUTPUT (fixed)
# -------------------------
def print_transactions_table(transactions):
    print(
        f"{'Date':<12} {'Description':<25} {'Amount':<10} "
        f"{'Category':<12} {'Large?':<8} {'This Month?'}"
    )
    print("-" * 85)

    for t in transactions:
        amount_str = f"{t.amount:.2f}"

        print(
            f"{t.date.strftime('%Y-%m-%d'):<12} "
            f"{t.description:<25} "
            f"{amount_str:<10} "
            f"{t.category:<12} "
            f"{'YES' if t.is_large_expense() else 'NO':<8} "
            f"{'YES' if t.is_this_month() else 'NO'}"
        )


# -------------------------
# 5. Run
# -------------------------
print_transactions_table(transactions)
````

````python
# Here’s a complete, clean Transaction Manager system in one block of code that includes:
# Adding transactions
# Total expenses
# Grouping by category
# Top purchases
# Filtering by month
# Simple usage example



from collections import defaultdict
from datetime import datetime


class Transaction:
    def __init__(self, date, amount, category, description=""):
        """
        date: string in format "YYYY-MM-DD"
        amount: positive = expense, negative = income/refund
        category: e.g. Food, Rent, Travel
        description: optional note
        """
        self.date = date
        self.amount = amount
        self.category = category
        self.description = description

    def is_expense(self):
        return self.amount > 0

    def __repr__(self):
        return f"{self.date} | {self.category} | {self.amount} | {self.description}"


class TransactionManager:
    def __init__(self):
        self.transactions = []

    # -------------------------
    # Add transaction
    # -------------------------
    def add_transaction(self, transaction):
        self.transactions.append(transaction)

    # -------------------------
    # Total expenses
    # -------------------------
    def total_expenses(self):
        return sum(t.amount for t in self.transactions if t.amount > 0)

    # -------------------------
    # Total income
    # -------------------------
    def total_income(self):
        return sum(-t.amount for t in self.transactions if t.amount < 0)

    # -------------------------
    # Group by category
    # -------------------------
    def expenses_by_category(self):
        category_totals = defaultdict(float)

        for t in self.transactions:
            if t.amount > 0:
                category_totals[t.category] += t.amount

        return dict(category_totals)

    # -------------------------
    # Top N biggest purchases
    # -------------------------
    def top_purchases(self, n=5):
        expenses = [t for t in self.transactions if t.amount > 0]
        expenses.sort(key=lambda x: x.amount, reverse=True)
        return expenses[:n]

    # -------------------------
    # Filter transactions for current month
    # -------------------------
    def this_month(self):
        now = datetime.now()
        results = []

        for t in self.transactions:
            t_date = datetime.strptime(t.date, "%Y-%m-%d")
            if t_date.month == now.month and t_date.year == now.year:
                results.append(t)

        return results

    # -------------------------
    # Monthly summary report
    # -------------------------
    def monthly_report(self):
        report = {
            "total_expenses": self.total_expenses(),
            "total_income": self.total_income(),
            "by_category": self.expenses_by_category(),
            "top_purchases": self.top_purchases()
        }
        return report


# =========================
# Example usage
# =========================
if __name__ == "__main__":
    tm = TransactionManager()

    tm.add_transaction(Transaction("2026-05-01", 120, "Food", "Groceries"))
    tm.add_transaction(Transaction("2026-05-03", 2000, "Rent", "Monthly rent"))
    tm.add_transaction(Transaction("2026-05-05", 50, "Transport", "Uber rides"))
    tm.add_transaction(Transaction("2026-05-06", -500, "Salary", "Part-time job"))

    print("\nTOTAL EXPENSES:", tm.total_expenses())
    print("TOTAL INCOME:", tm.total_income())

    print("\nEXPENSES BY CATEGORY:")
    print(tm.expenses_by_category())

    print("\nTOP PURCHASES:")
    for t in tm.top_purchases():
        print(t)

    print("\nTHIS MONTH TRANSACTIONS:")
    for t in tm.this_month():
        print(t)

    print("\nFULL REPORT:")
    print(tm.monthly_report())
````

````python
import pandas as pd
from datetime import datetime


class Transaction:
    def __init__(self, date, amount, category, description=""):
        self.date = date
        self.amount = amount
        self.category = category
        self.description = description

    def to_dict(self):
        return {
            "Date": self.date,
            "Category": self.category,
            "Amount": self.amount,
            "Description": self.description
        }


class TransactionManager:
    def __init__(self):
        self.transactions = []

    # -------------------------
    # Add transaction
    # -------------------------
    def add_transaction(self, transaction):
        self.transactions.append(transaction)

    # -------------------------
    # Convert to DataFrame
    # -------------------------
    def to_df(self):
        return pd.DataFrame([t.to_dict() for t in self.transactions])

    # -------------------------
    # Summary Table
    # -------------------------
    def summary_table(self):
        df = self.to_df()

        expenses = df[df["Amount"] > 0]["Amount"].sum()
        income = -df[df["Amount"] < 0]["Amount"].sum()
        net = income - expenses

        summary = pd.DataFrame({
            "Metric": ["Total Expenses", "Total Income", "Net Cash Flow"],
            "Amount ($)": [expenses, income, net]
        })

        return summary

    # -------------------------
    # Expenses by category
    # -------------------------
    def category_table(self):
        df = self.to_df()
        expenses = df[df["Amount"] > 0]

        category = expenses.groupby("Category")["Amount"].sum().reset_index()
        category.columns = ["Category", "Total Spent ($)"]

        return category.sort_values(by="Total Spent ($)", ascending=False)

    # -------------------------
    # Top purchases
    # -------------------------
    def top_purchases(self, n=5):
        df = self.to_df()
        expenses = df[df["Amount"] > 0]

        top = expenses.sort_values(by="Amount", ascending=False).head(n)
        return top.reset_index(drop=True)

    # -------------------------
    # Monthly filter
    # -------------------------
    def this_month(self):
        df = self.to_df()

        now = datetime.now()

        df["Date"] = pd.to_datetime(df["Date"])

        filtered = df[
            (df["Date"].dt.month == now.month) &
            (df["Date"].dt.year == now.year)
        ]

        return filtered.reset_index(drop=True)

    # -------------------------
    # Full report
    # -------------------------
    def full_report(self):
        print("\n=== SUMMARY ===")
        print(self.summary_table().to_string(index=False))

        print("\n=== EXPENSES BY CATEGORY ===")
        print(self.category_table().to_string(index=False))

        print("\n=== TOP PURCHASES ===")
        print(self.top_purchases().to_string(index=False))

        print("\n=== THIS MONTH ===")
        print(self.this_month().to_string(index=False))


# =========================
# Example usage
# =========================
if __name__ == "__main__":
    tm = TransactionManager()

    tm.add_transaction(Transaction("2026-05-01", 120, "Food", "Groceries"))
    tm.add_transaction(Transaction("2026-05-03", 2000, "Rent", "Monthly rent"))
    tm.add_transaction(Transaction("2026-05-05", 50, "Transport", "Uber rides"))
    tm.add_transaction(Transaction("2026-05-06", -500, "Salary", "Part-time job"))

    tm.full_report()
````
