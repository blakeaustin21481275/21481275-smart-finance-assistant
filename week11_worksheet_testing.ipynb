## Exercise 5.1: Specification-First Bug Fixing

You discover this buggy code in production:

````python
def average_positive(numbers):
    """Calculate average of positive numbers only."""
    total = 0
    count = 0
    for num in numbers:
        if num > 0:
            total += num
        count += 1
    return total / count
````

**Step 1: Write a failing test that demonstrates the bug**

````python
# First, understand what SHOULD happen:
# Input: [2, -1, 4, -3]
# Positive numbers: [2, 4]
# Average: (2 + 4) / 2 = 3.0

# Write the assertion that currently fails:
def average_positive(numbers):
    """Calculate average of positive numbers only."""
    total = 0
    count = 0
    for num in numbers:
        if num > 0:
            total += num
        count += 1
    return total / count


# Problems with code: count outside of num > 0 (counts all numbers), doesn't account for dividing 0


# TODO: Write assertion here that will fail
assert average_positive([2, -1, 4, -3]) == 3
````

**Step 2: Specify the bug to AI**

Write a clear bug report prompt for AI:

````python
"""
Your prompt to AI:

The function average_positive should calculate the average of ONLY positive numbers,
ignoring negative numbers.

Current behavior:
AssertionError, test fails, likely due to faulty code

Expected behavior:
Should output a value of 3.0 ie (2+4)/2 = 3.0

Here are the tests that should pass:
assert average_positive([2, -1, 4, -3]) == 3.0
assert average_positive([-1 -3]) == 0



Please fix the function to pass these tests. Keep it simple.
"""
````

**Step 3: Get AI's fix and verify**

````python
# Paste AI's corrected version here:
def average_positive(numbers):
    """Calculate average of positive numbers only."""

    total = 0
    count = 0

    for num in numbers:
        if num > 0:
            total += num
            count += 1

    if count == 0:
        return 0

    return total / count


# Tests
assert average_positive([2, -1, 4, -3]) == 3.0
assert average_positive([-1, -3]) == 0

print("All tests passed!")
````

**Step 4: Add edge case tests**

Now that the basic bug is fixed, what edge cases should you add?

````python
"""
Additional scenarios to test:

1. All negative numbers: [-1, -2, -3]
   Expected behavior:0

2. Empty list: []
   Expected behavior: return error message (no values)

3. Mix with zero: [0, 1, 2]
   Expected behavior:3/3 = 1
"""

# Write assertions for these edge cases:

# 1. All negative numbers
assert average_positive([-1, -2, -3]) == 0

# 2. Empty list
try:
    average_positive([])
    assert False  # should not reach here
except ValueError:
    assert True

# 3. Mix with zero
assert average_positive([0, 1, 2]) == 1
````

````python
def average_positive(numbers):
    """Calculate average of positive numbers only, with edge cases handled."""

    if len(numbers) == 0:
        return "Error: no values"

    total = 0
    count = 0

    for num in numbers:
        if num > 0:
            total += num
            count += 1

    if count == 0:
        return 0

    return total / count


# Tests
assert average_positive([2, -1, 4, -3]) == 3.0
assert average_positive([-1, -2, -3]) == 0
assert average_positive([]) == "Error: no values"
assert average_positive([0, 1, 2]) == 1.5 # 0 isn't counted as a positive number

print("All tests passed!")
````

````python
# Paste AI's code:


def calculate_grade(score):
    """Calculate letter grade from numeric score."""

    # Handle invalid input types
    if not isinstance(score, (int, float)):
        return "Invalid input"

    # Handle out-of-range scores
    if score < 0 or score > 100:
        return "Invalid input"

    # Grade calculation
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"


# Standard cases
assert calculate_grade(95) == "A"
assert calculate_grade(85) == "B"
assert calculate_grade(75) == "C"
assert calculate_grade(65) == "D"
assert calculate_grade(55) == "F"

# Boundary cases
assert calculate_grade(90) == "A"
assert calculate_grade(80) == "B"
assert calculate_grade(70) == "C"
assert calculate_grade(60) == "D"

# Just above boundaries
assert calculate_grade(91) == "A"
assert calculate_grade(81) == "B"
assert calculate_grade(71) == "C"
assert calculate_grade(61) == "D"

# Upper values in ranges
assert calculate_grade(99) == "A"
assert calculate_grade(89) == "B"
assert calculate_grade(79) == "C"
assert calculate_grade(69) == "D"
assert calculate_grade(59) == "F"

# Additional edge cases
assert calculate_grade(0) == "F"
assert calculate_grade(100) == "A"

# Invalid inputs
assert calculate_grade(-5) == "Invalid input"
assert calculate_grade(105) == "Invalid input"
assert calculate_grade("90") == "Invalid input"
assert calculate_grade(None) == "Invalid input"

print("All tests passed!")
````
