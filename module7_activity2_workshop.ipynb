````python
# Your AI collaboration code goes here
# Try the prompt above and see what AI suggests!


import pandas as pd

# Load your data
df = pd.read_csv('transactions.csv')

# Clean Amount column (remove $ and convert to float)
df['Amount'] = pd.to_numeric(df['Amount'].astype(str).str.replace('$', ''), errors='coerce')

# Group and analyze
spending_analysis = df.groupby('Category').agg(
    Total_Spent=('Amount', 'sum'),
    Avg_Amount=('Amount', 'mean'),
    Transaction_Count=('Amount', 'count')
).round(2)

# Sort by highest spending
spending_analysis = spending_analysis.sort_values('Total_Spent', ascending=False)

# Calculate overall metrics
total_spending = spending_analysis['Total_Spent'].sum()
avg_spending = spending_analysis['Total_Spent'].mean()

print("💼 Spending Summary & Savings Opportunities")
print("=" * 60)

# Identify potential savings categories
for category, row in spending_analysis.iterrows():
    total = row['Total_Spent']
    avg = row['Avg_Amount']
    count = row['Transaction_Count']

    flags = []

    # Flag high total spending
    if total > avg_spending:
        flags.append("High Total Spending")

    # Flag frequent transactions
    if count > spending_analysis['Transaction_Count'].mean():
        flags.append("Frequent Transactions")

    # Flag high average transaction
    if avg > spending_analysis['Avg_Amount'].mean():
        flags.append("High Avg Transaction")

    flag_text = ", ".join(flags) if flags else "OK"

    print(f"{category:<15} Total: ${total:<10.2f} Avg: ${avg:<10.2f} Count: {count:<5} → {flag_text}")

print("\n📊 Key Insights:")
print("- Categories flagged may offer opportunities to cut costs.")
print("- Frequent small purchases can often be reduced with budgeting.")
print("- High average expenses may be negotiable or replaceable.")

print(f"\n💰 Grand Total Spending: ${total_spending:.2f}")
````

## 📊 Step 1: Set Up Your Data

**Open your assignment notebook** (`smart_finance_assistant.ipynb`) and start with this section:

### Data Loading and Exploration

::: {.callout-tip}
## Prompt AI for this section:

*"I'm building a Smart Finance Assistant for my programming assignment. I need to load a CSV file with transaction data (columns: Date, Amount, Category, Description). The Amount column has dollar signs that need cleaning. Please write professional Python code with detailed comments that:
1. Loads the CSV data
2. Cleans and validates the data
3. Shows basic information about the dataset

````python
# AI-assisted data loading section
# (Copy the code AI provides here)

"""
Smart Finance Assistant - Data Loader & Cleaner

This script:
1. Loads transaction data from a CSV file
2. Cleans and validates the dataset
3. Displays useful summary information
4. Handles common data issues gracefully
"""

import pandas as pd
import numpy as np
import logging

# -------------------------------
# CONFIGURATION
# -------------------------------

# Set up logging for better debugging and traceability
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')

# File path (update as needed)
FILE_PATH = "transactions.csv"

# Expected columns
EXPECTED_COLUMNS = ["Date", "Amount", "Category", "Description"]


# -------------------------------
# LOAD DATA FUNCTION
# -------------------------------

def load_data(file_path):
    """
    Load CSV data into a pandas DataFrame.

    Parameters:
        file_path (str): Path to the CSV file

    Returns:
        pd.DataFrame: Loaded DataFrame
    """
    try:
        df = pd.read_csv(file_path)
        logging.info("CSV file loaded successfully.")
        return df
    except FileNotFoundError:
        logging.error(f"File not found: {file_path}")
        raise
    except pd.errors.EmptyDataError:
        logging.error("The file is empty.")
        raise
    except Exception as e:
        logging.error(f"Unexpected error while loading file: {e}")
        raise


# -------------------------------
# VALIDATE STRUCTURE
# -------------------------------

def validate_columns(df):
    """
    Ensure required columns exist in the dataset.

    Parameters:
        df (pd.DataFrame): Input DataFrame

    Raises:
        ValueError: If required columns are missing
    """
    missing_cols = [col for col in EXPECTED_COLUMNS if col not in df.columns]

    if missing_cols:
        raise ValueError(f"Missing required columns: {missing_cols}")

    logging.info("All required columns are present.")


# -------------------------------
# CLEAN DATA
# -------------------------------

def clean_data(df):
    """
    Clean and preprocess the dataset.

    Steps:
    - Strip whitespace from string columns
    - Convert Date column to datetime
    - Clean Amount column (remove $ and commas)
    - Handle missing values

    Parameters:
        df (pd.DataFrame): Raw DataFrame

    Returns:
        pd.DataFrame: Cleaned DataFrame
    """

    # Create a copy to avoid modifying original data
    df = df.copy()

    # ---------------------------
    # Clean string columns
    # ---------------------------
    for col in ["Category", "Description"]:
        df[col] = df[col].astype(str).str.strip()

    # ---------------------------
    # Clean Date column
    # ---------------------------
    df["Date"] = pd.to_datetime(df["Date"], errors='coerce')

    # ---------------------------
    # Clean Amount column
    # ---------------------------
    # Remove dollar signs and commas, then convert to float
    df["Amount"] = (
        df["Amount"]
        .astype(str)
        .str.replace(r"[\$,]", "", regex=True)
    )

    df["Amount"] = pd.to_numeric(df["Amount"], errors='coerce')

    # ---------------------------
    # Handle missing values
    # ---------------------------
    missing_before = df.isnull().sum()

    # Drop rows where critical fields are missing
    df.dropna(subset=["Date", "Amount"], inplace=True)

    # Fill non-critical missing values
    df["Category"].fillna("Unknown", inplace=True)
    df["Description"].fillna("No Description", inplace=True)

    missing_after = df.isnull().sum()

    logging.info("Missing values before cleaning:\n%s", missing_before)
    logging.info("Missing values after cleaning:\n%s", missing_after)

    return df


# -------------------------------
# DATASET SUMMARY
# -------------------------------

def show_basic_info(df):
    """
    Display useful information about the dataset.

    Parameters:
        df (pd.DataFrame): Cleaned DataFrame
    """

    print("\n--- DATASET OVERVIEW ---")
    print(f"Number of rows: {df.shape[0]}")
    print(f"Number of columns: {df.shape[1]}")

    print("\n--- COLUMN TYPES ---")
    print(df.dtypes)

    print("\n--- SAMPLE DATA ---")
    print(df.head())

    print("\n--- STATISTICAL SUMMARY (Amount) ---")
    print(df["Amount"].describe())

    print("\n--- DATE RANGE ---")
    print(f"From: {df['Date'].min()} To: {df['Date'].max()}")

    print("\n--- CATEGORY DISTRIBUTION ---")
    print(df["Category"].value_counts())


# -------------------------------
# MAIN EXECUTION
# -------------------------------

def main():
    """
    Main execution pipeline.
    """

    # Step 1: Load data
    df = load_data(FILE_PATH)

    # Step 2: Validate structure
    validate_columns(df)

    # Step 3: Clean data
    df_clean = clean_data(df)

    # Step 4: Display dataset info
    show_basic_info(df_clean)

    logging.info("Data processing completed successfully.")


# Run script
if __name__ == "__main__":
    main()


# Add your own comments about what you learned
````

````python
# AI-generated analysis functions
# (Your AI collaboration code goes here)

# 1. Total Spending By Category
def total_spending_by_category(df):
    """
    BUSINESS PURPOSE:
    -----------------
    Helps users understand where their money is going by summarizing
    total spending per category (e.g., Food, Transport, Bills).

    PARAMETERS:
    -----------
    df (pd.DataFrame): Cleaned transaction dataset

    RETURNS:
    --------
    pd.Series: Total spending grouped by category (descending)

    RAISES:
    -------
    ValueError: If required columns are missing or dataset is empty
    """

    required_cols = {"Category", "Amount"}

    # Validate input
    if df is None or df.empty:
        raise ValueError("Dataset is empty or not loaded.")

    if not required_cols.issubset(df.columns):
        raise ValueError(f"Missing required columns: {required_cols}")

    try:
        result = (
            df.groupby("Category")["Amount"]
            .sum()
            .sort_values(ascending=False)
        )

        return result

    except Exception as e:
        raise RuntimeError(f"Error calculating spending by category: {e}")

# 2. Top 5 Spending Categories
def top_5_categories(df):
    """
    BUSINESS PURPOSE:
    -----------------
    Identifies the user's highest spending categories to highlight
    major cost drivers and potential savings opportunities.

    PARAMETERS:
    -----------
    df (pd.DataFrame): Cleaned transaction dataset

    RETURNS:
    --------
    pd.Series: Top 5 categories by total spending

    RAISES:
    -------
    ValueError: If dataset is invalid
    """

    try:
        category_totals = total_spending_by_category(df)

        top5 = category_totals.head(5)

        return top5

    except Exception as e:
        raise RuntimeError(f"Error finding top 5 categories: {e}")


# 3. Average Spending per Transaction
def average_transaction_amount(df):
    """
    BUSINESS PURPOSE:
    -----------------
    Provides insight into typical spending behavior by calculating
    the average transaction value.

    PARAMETERS:
    -----------
    df (pd.DataFrame): Cleaned transaction dataset

    RETURNS:
    --------
    float: Average transaction amount

    RAISES:
    -------
    ValueError: If dataset is invalid or contains no valid amounts
    """

    if df is None or df.empty:
        raise ValueError("Dataset is empty.")

    if "Amount" not in df.columns:
        raise ValueError("Missing 'Amount' column.")

    try:
        avg = df["Amount"].mean()

        if np.isnan(avg):
            raise ValueError("No valid numeric transaction data.")

        return round(avg, 2)

    except Exception as e:
        raise RuntimeError(f"Error calculating average transaction: {e}")

# 4. Identify Unusual Transactions
def detect_unusual_transactions(df, threshold_multiplier=3):
    """
    BUSINESS PURPOSE:
    -----------------
    Flags potentially suspicious or noteworthy transactions:
    - Extremely high spending (outliers)
    - Negative values (possible refunds or errors)

    PARAMETERS:
    -----------
    df (pd.DataFrame): Cleaned transaction dataset
    threshold_multiplier (float): Sensitivity for outlier detection
                                  (default = 3 standard deviations)

    RETURNS:
    --------
    pd.DataFrame: Subset of unusual transactions

    RAISES:
    -------
    ValueError: If dataset is invalid
    """

    if df is None or df.empty:
        raise ValueError("Dataset is empty.")

    if "Amount" not in df.columns:
        raise ValueError("Missing 'Amount' column.")

    try:
        mean = df["Amount"].mean()
        std = df["Amount"].std()

        # Define thresholds
        upper_limit = mean + (threshold_multiplier * std)
        lower_limit = mean - (threshold_multiplier * std)

        unusual = df[
            (df["Amount"] > upper_limit) |   # very high transactions
            (df["Amount"] < lower_limit) |   # unusually low
            (df["Amount"] < 0)               # negative transactions
        ]

        return unusual.sort_values(by="Amount", ascending=False)

    except Exception as e:
        raise RuntimeError(f"Error detecting unusual transactions: {e}")

````

## 💰 Step 3: Generate Business Insights

**Challenge AI** to help you create business-relevant insights:

::: {.callout-tip}
## Business Insights Prompt:

*"Using the transaction analysis functions, help me create a comprehensive spending report that would be useful for personal financial planning. Include recommendations for potential savings and spending pattern observations. Format it like a professional financial summary."*
:::

````python
# AI-generated business insights
# (Your AI collaboration code goes here)

import pandas as pd
import numpy as np


# -------------------------------
# REPORT GENERATOR FUNCTION
# -------------------------------

def generate_spending_report(df):
    """
    Generates a professional financial spending report.

    PARAMETERS:
    -----------
    df (pd.DataFrame): Cleaned transaction dataset

    RETURNS:
    --------
    str: Formatted financial report
    """

    if df is None or df.empty:
        raise ValueError("Dataset is empty. Cannot generate report.")

    required_cols = {"Date", "Amount", "Category"}
    if not required_cols.issubset(df.columns):
        raise ValueError(f"Missing required columns: {required_cols}")

    # -------------------------------
    # BASIC METRICS
    # -------------------------------
    total_transactions = len(df)
    avg_transaction = df["Amount"].mean()
    total_spent = df["Amount"].sum()

    date_range = f"{df['Date'].min().date()} to {df['Date'].max().date()}"

    # -------------------------------
    # CATEGORY ANALYSIS
    # -------------------------------
    category_totals = df.groupby("Category")["Amount"].sum().sort_values(ascending=False)
    top_categories = category_totals.head(5)

    # Calculate percentages
    category_percent = (category_totals / total_spent * 100).round(2)

    # -------------------------------
    # UNUSUAL TRANSACTIONS
    # -------------------------------
    mean = df["Amount"].mean()
    std = df["Amount"].std()

    upper_limit = mean + (3 * std)
    lower_limit = mean - (3 * std)

    unusual = df[
        (df["Amount"] > upper_limit) |
        (df["Amount"] < lower_limit) |
        (df["Amount"] < 0)
    ]

    # -------------------------------
    # BUILD REPORT STRING
    # -------------------------------

    report = []
    report.append("=" * 50)
    report.append("📊 PERSONAL SPENDING REPORT")
    report.append("=" * 50)

    # Executive Summary
    report.append("\n🔹 EXECUTIVE SUMMARY")
    report.append(f"Period: {date_range}")
    report.append(f"Total Transactions: {total_transactions}")
    report.append(f"Total Spending: ${total_spent:,.2f}")
    report.append(f"Average Transaction: ${avg_transaction:.2f}")
    report.append(f"Top Category: {category_totals.idxmax()}")

    # Category Breakdown
    report.append("\n🔹 SPENDING BY CATEGORY")
    for cat, value in category_totals.items():
        percent = category_percent[cat]
        report.append(f"{cat:<15} ${value:>10,.2f} ({percent}%)")

    # Top Categories
    report.append("\n🔹 TOP 5 CATEGORIES")
    for i, (cat, value) in enumerate(top_categories.items(), 1):
        report.append(f"{i}. {cat} - ${value:,.2f}")

    # Unusual Transactions
    report.append("\n🔹 UNUSUAL TRANSACTIONS")
    if unusual.empty:
        report.append("No unusual transactions detected.")
    else:
        report.append(f"{len(unusual)} unusual transactions found:")
        report.append(unusual[["Date", "Category", "Amount"]].to_string(index=False))

    # -------------------------------
    # INSIGHTS & RECOMMENDATIONS
    # -------------------------------

    report.append("\n🔹 KEY INSIGHTS")

    # Insight 1: Spending concentration
    top_3_share = category_totals.head(3).sum() / total_spent * 100
    report.append(f"- Top 3 categories account for {top_3_share:.1f}% of total spending.")

    # Insight 2: High spending detection
    if avg_transaction > 100:
        report.append("- Average transaction value is relatively high.")
    else:
        report.append("- Spending consists of many smaller transactions.")

    # -------------------------------
    # RECOMMENDATIONS
    # -------------------------------

    report.append("\n🔹 RECOMMENDATIONS")

    # Recommendation 1
    top_category = category_totals.idxmax()
    report.append(f"- Reduce spending in '{top_category}' by 5–10% to save money.")

    # Recommendation 2
    report.append("- Set category-based monthly budgets to control expenses.")

    # Recommendation 3
    if not unusual.empty:
        report.append("- Review unusual transactions for errors or avoidable expenses.")

    # Recommendation 4
    report.append("- Track daily spending to reduce unnoticed small expenses.")

    report.append("\n" + "=" * 50)

    return "\n".join(report)


# -------------------------------
# EXAMPLE USAGE
# -------------------------------
if __name__ == "__main__":
    # Example: Load your cleaned dataset
    df = pd.read_csv("cleaned_transactions.csv", parse_dates=["Date"])

    report = generate_spending_report(df)

    print(report)
````
